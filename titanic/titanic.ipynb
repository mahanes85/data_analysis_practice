```python
male = (df["sex"] == "male").sum()
female = (df["sex"] == "female").sum()

print(f"male: {male}, female: {female}")
```

```python
male = 0
female = 0

for survived, sex in zip(df["survived"], df["sex"]):
    if survived == 1:
        if sex == "male":
            male += 1
        else:
            female += 1

print(f"male: {male}, female: {female}")
```

```python
pclass_3 = 0
pclass_2 = 0
pclass_1 = 0

for survived, pclass in zip(df["survived"], df["pclass"]):
    if survived == 1:
        if pclass == 3:
            pclass_3 += 1
        elif pclass == 2:
            pclass_2 += 1
        elif pclass == 1:
            pclass_1 += 1

print(f"pclass_3: {pclass_3}, pclass_2: {pclass_2}, pclass_1 : {pclass_1}")
```

```python
under_10 = 0
under_20 = 0
more_than_20 =0

for survived, age in zip(df["survived"], df["age"]):
    if survived == 1:
        if age <= 10:
            under_10 += 1
        elif 10 < age <= 20:
            under_20 += 1
        elif age > 20:
            more_than_20 += 1

print(f"under_10: {under_10}, under_20: {under_20}, more_than_20 : {more_than_20}")
```
