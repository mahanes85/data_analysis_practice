# titanic_data

## Load Library

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data

```python
df = sns.load_dataset("titanic")
df
```

## Plot

```python
plt.hist(df["survived"], bins=2)
```

```python
plt.figure(figsize=(15,6))

plt.subplot(121)
plt.title("men")
plt.hist(df[df["who"]=="man"]["survived"], bins=2)

plt.subplot(122)
plt.title("women")
plt.hist(df[df["who"]=="woman"]["survived"], bins=2)
```

```python
plt.figure(figsize=(15,6))

plt.subplot(131)
plt.title("First_Class_man")
plt.hist(df[(df["who"]=="man") & (df["class"]=="First")]["survived"], bins=2)

plt.subplot(132)
plt.title("Second_Class_man")
plt.hist(df[(df["who"]=="man") & (df["class"]=="Second")]["survived"], bins=2)

plt.subplot(133)
plt.title("Third_Class_man")
plt.hist(df[(df["who"]=="man") & (df["class"]=="Third")]["survived"], bins=2)
```

```python
plt.figure(figsize=(15,6))

plt.subplot(131)
plt.title("First_Class_woman")
plt.hist(df[(df["who"]=="woman") & (df["class"]=="First")]["survived"], bins=2)

plt.subplot(132)
plt.title("Second_Class_woman")
plt.hist(df[(df["who"]=="woman") & (df["class"]=="Second")]["survived"], bins=2)

plt.subplot(133)
plt.title("Third_Class_woman")
plt.hist(df[(df["who"]=="woman") & (df["class"]=="Third")]["survived"], bins=2)
```

```python
cities = list(df["embark_town"].value_counts().index)

for city in cities:
    plt.hist(df[(df["who"]=="man") & (df["class"]=="Third") & (df["embark_town"]==city)]["survived"], bins=2)
    plt.show()
```

```python
cities = list(df["embark_town"].value_counts().index)

for city in cities:
    plt.hist(df[(df["who"]=="man") & (df["class"]=="Second") & (df["embark_town"]==city)]["survived"], bins=2)
    plt.show()
```
