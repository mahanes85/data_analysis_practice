## Load Library

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data

```python
df = sns.load_dataset("taxis")
df
```

```python
df.info()
```

```python
df.columns
```

```python
df.index
```

## Plot

```python
plt.scatter(df["passengers"],df["fare"])
```

```python
plt.scatter(df["passengers"],df["tip"])
```

```python
plt.scatter(df["color"],df["fare"])
```

```python
df["payment"] = df["payment"].map({"credit card":1, "cash":0})
plt.scatter(df["payment"],df["tip"])
```

```python
plt.scatter(df["distance"],df["fare"])
```

```python
plt.scatter(df["distance"],df["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Yellow")
plt.hist(df[(df["payment"]==0) & (df["color"]=="yellow")]["tip"])
plt.subplot(122)
plt.title("Green")
plt.hist(df[(df["payment"]==0) & (df["color"]=="green")]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Yellow")
plt.hist(df[(df["payment"]==1) & (df["color"]=="yellow")]["tip"])
plt.subplot(122)
plt.title("Green")
plt.hist(df[(df["payment"]==1) & (df["color"]=="green")]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Yellow")
plt.hist(df[(df["passengers"]>=3) & (df["color"]=="yellow")]["tip"])
plt.subplot(122)
plt.title("Green")
plt.hist(df[(df["passengers"]>=3) & (df["color"]=="green")]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Yellow")
plt.hist(df[(df["distance"]>=10) & (df["color"]=="yellow")]["fare"])
plt.subplot(122)
plt.title("Green")
plt.hist(df[(df["distance"]>=10) & (df["color"]=="green")]["fare"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Yellow")
plt.hist(df[(df["distance"]>=10) & (df["passengers"]>=3) & (df["color"]=="yellow")]["fare"])
plt.subplot(122)
plt.title("Green")
plt.hist(df[(df["distance"]>=10) & (df["passengers"]>=3) & (df["color"]=="green")]["fare"])
```
