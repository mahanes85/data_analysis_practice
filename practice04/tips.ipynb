## Load Library

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data

```python
df = sns.load_dataset("tips")
df
```

## Plot

```python
plt.scatter(df["total_bill"],df["tip"])
```

```python
plt.scatter(df["sex"],df["total_bill"])
```

```python
plt.scatter(df["sex"],df["tip"])
```

```python
plt.scatter(df["time"],df["total_bill"])
```

```python
plt.scatter(df["size"],df["total_bill"])
```

```python
plt.scatter(df["size"],df["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Male")
plt.hist(df[(df["sex"]=="Male") & (df["total_bill"] >= 30)]["tip"])
plt.subplot(122)
plt.title("Female")
plt.hist(df[(df["sex"]=="Female") & (df["total_bill"] >= 30)]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Smoker")
plt.hist(df[(df["sex"]=="Male") & (df["smoker"]=='Yes')]["tip"])
plt.subplot(122)
plt.title("Non smoker")
plt.hist(df[(df["sex"]=="Male") & (df["smoker"]=='No')]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Smoker")
plt.hist(df[(df["sex"]=="Female") & (df["smoker"]=='Yes')]["tip"])
plt.subplot(122)
plt.title("Non smoker")
plt.hist(df[(df["sex"]=="Female") & (df["smoker"]=='No')]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Lunch")
plt.hist(df[(df["time"]=="Lunch") & (df["size"]>=3)]["tip"])
plt.subplot(122)
plt.title("Dinner")
plt.hist(df[(df["time"]=="Dinner") & (df["size"]>=3)]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Lunch")
plt.hist(df[(df["time"]=="Lunch") & (df["size"]<3)]["tip"])
plt.subplot(122)
plt.title("Dinner")
plt.hist(df[(df["time"]=="Dinner") & (df["size"]<3)]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Male group")
plt.hist(df[(df["time"]=="Lunch") & (df["sex"]=="Male") & (df["size"]>=3)]["tip"])
plt.subplot(122)
plt.title("Female group")
plt.hist(df[(df["time"]=="Lunch") & (df["sex"]=="Female") & (df["size"]>=3)]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Male group")
plt.hist(df[(df["time"]=="Dinner") & (df["sex"]=="Male") & (df["size"]>=3)]["tip"])
plt.subplot(122)
plt.title("Female group")
plt.hist(df[(df["time"]=="Dinner") & (df["sex"]=="Female") & (df["size"]>=3)]["tip"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(121)
plt.title("Male group")
plt.hist(df[(df["time"]=="Lunch") & (df["sex"]=="Male") & (df["size"]>=3)]["total_bill"])
plt.subplot(122)
plt.title("Female group")
plt.hist(df[(df["time"]=="Lunch") & (df["sex"]=="Female") & (df["size"]>=3)]["total_bill"])
```
