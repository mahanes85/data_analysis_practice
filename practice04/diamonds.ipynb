```python
plt.figure(figsize=(15,5))

plt.subplot(131)
plt.title("Premium")
plt.hist(df[(df["carat"] <= 1) & (df["cut"]=="Premium")]["price"])
plt.subplot(132)
plt.title("Very Good")
plt.hist(df[(df["carat"] <= 1) & (df["cut"]=="Very Good")]["price"])
plt.subplot(133)
plt.title("Good")
plt.hist(df[(df["carat"] <= 1) & (df["cut"]=="Good")]["price"])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(131)
plt.title("I1_worst")
plt.hist(df[(df["carat"] <= 1) & (df["clarity"]=="I1")]["price"])
plt.subplot(132)
plt.title("VS1_average")
plt.hist(df[(df["carat"] <= 1) & (df["clarity"]=="VS1")]["price"])
plt.subplot(133)
plt.title("IF_best")
plt.hist(df[(df["carat"] <= 1) & (df["clarity"]=="IF")]["price"])
```
