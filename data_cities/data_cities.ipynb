# data_cities

## Load Library

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Data

```python
Data = pd.read_csv("data_cities.csv")
Data
```

## Plotting a graph

```python
plt.scatter(Data["longitude"], Data["latitude"], marker="p")

plt.title("Map Of Iran")
plt.xlabel("longitude")
plt.ylabel("latitude")
plt.xlim(40,65)
plt.ylim(25,40)
plt.grid()
plt.show()
```
